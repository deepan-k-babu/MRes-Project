```python
from skbio.stats.distance import permanova
from skbio.diversity import beta_diversity
import pandas as pd
from itertools import combinations

# Ensure group_labels has a name
cultivar_labels.name = 'Group'

# Filter df_order to match group_labels
df_order = df_order.loc[group_labels.index]

# Compute Bray-Curtis distance matrix
dm = beta_diversity(
    metric='braycurtis',
    counts=df_order.values,
    ids=df_order.index
)

# Get unique group names
groups = group_labels.unique()

# Run pairwise PERMANOVA
results = []
for g1, g2 in combinations(groups, 2):
    selected_ids = group_labels[(group_labels == g1) | (group_labels == g2)].index
    sub_dm = dm.filter(selected_ids)
    sub_labels = group_labels.loc[selected_ids]
    
    result = permanova(distance_matrix=sub_dm, grouping=sub_labels)
    result_dict = result.to_dict()
    result_dict['Group 1'] = g1
    result_dict['Group 2'] = g2
    results.append(result_dict)

# Convert results to DataFrame
pairwise_df = pd.DataFrame(results)

# Apply Bonferroni correction
num_tests = len(pairwise_df)
pairwise_df['Bonferroni corrected p-value'] = (pairwise_df['p-value'] * num_tests).clip(upper=1.0)

# Display results
print(pairwise_df[['Group 1', 'Group 2', 'test statistic', 'p-value', 'Bonferroni corrected p-value']])
```

```python
from skbio.stats.distance import permanova
from skbio.diversity import beta_diversity
import pandas as pd
from itertools import combinations
from statsmodels.stats.multitest import multipletests

# Ensure group_labels has a name
group_labels.name = 'Group'

# Filter df_order to match group_labels
df_order = df_order.loc[group_labels.index]

# Compute Bray-Curtis distance matrix
dm = beta_diversity(
    metric='braycurtis',
    counts=df_order.values,
    ids=df_order.index
)

# Get unique group names
groups = group_labels.unique()

# Run pairwise PERMANOVA
results = []
p_values = []

for g1, g2 in combinations(groups, 2):
    selected_ids = group_labels[(group_labels == g1) | (group_labels == g2)].index
    sub_dm = dm.filter(selected_ids)
    sub_labels = group_labels.loc[selected_ids]
    
    result = permanova(distance_matrix=sub_dm, grouping=sub_labels)
    result_dict = result.to_dict()
    result_dict['Group 1'] = g1
    result_dict['Group 2'] = g2
    results.append(result_dict)
    p_values.append(result_dict['p-value'])

# Apply FDR correction
_, fdr_corrected, _, _ = multipletests(p_values, method='fdr_bh')

# Add corrected p-values to results
for i, corrected in enumerate(fdr_corrected):
    results[i]['FDR corrected p-value'] = corrected

# Convert to DataFrame
pairwise_df = pd.DataFrame(results)
print(pairwise_df[['Group 1', 'Group 2', 'test statistic', 'p-value', 'FDR corrected p-value']])
```

```python
from skbio.stats.distance import permanova
from skbio.diversity import beta_diversity
from itertools import combinations
import pandas as pd
from statsmodels.stats.multitest import multipletests

# Choose which label to test: cultivar_labels or group_labels
labels = cultivar_labels.copy()
labels.name = 'Cultivar'

# Filter df_order to match labels
df_order = df_order.loc[labels.index]

# Compute Bray-Curtis distance matrix
dm = beta_diversity(
    metric='braycurtis',
    counts=df_order.values,
    ids=df_order.index
)

# Run pairwise PERMANOVA
results = []
groups = labels.unique()

for g1, g2 in combinations(groups, 2):
    selected_ids = labels[(labels == g1) | (labels == g2)].index
    sub_dm = dm.filter(selected_ids)
    sub_labels = labels.loc[selected_ids]
    
    result = permanova(distance_matrix=sub_dm, grouping=sub_labels)
    result_dict = result.to_dict()
    result_dict['Group 1'] = g1
    result_dict['Group 2'] = g2
    results.append(result_dict)

# Convert to DataFrame
pairwise_df = pd.DataFrame(results)

pvals = pairwise_df['p-value'].values
_, fdr_corrected, _, _ = multipletests(pvals, method='fdr_bh')
pairwise_df['FDR corrected p-value'] = fdr_corrected

pairwise_df
```

```python
from scipy.stats import kruskal
from statsmodels.stats.multitest import multipletests
import pandas as pd

# Ensure df_order and group_labels are aligned
df_order = df_order.loc[group_labels.index]

# Initialize results list
results = []

# Loop through each taxon (column)
for taxon in df_order.columns:
    groups = []
    for sub in group_labels.unique():
        group_data = df_order.loc[group_labels == sub, taxon]
        groups.append(group_data.dropna())
    
    # Run Kruskal-Wallis test if there are at least 2 groups
    if len(groups) >= 2:
        stat, p = kruskal(*groups)
        results.append({'Taxon': taxon, 'Statistic': stat, 'p-value': p})

# Convert to DataFrame
results_df = pd.DataFrame(results)

# Apply FDR correction
_, fdr_pvals, _, _ = multipletests(results_df['p-value'], method='fdr_bh')
results_df['FDR corrected p-value'] = fdr_pvals

# Filter significant taxa
significant = results_df[results_df['FDR corrected p-value'] < 0.05]
print(significant.sort_values('FDR corrected p-value'))
```
